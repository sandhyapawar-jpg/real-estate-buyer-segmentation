```python
import pandas as pd

df = pd.read_csv('/content/real_estate_buyer_segmentation_dataset (1).csv')

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
# Check the dataset
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Basic statistics for the dataset
df.describe(include='all')
```

```python
# Check for duplicate rows and missing values

print("Duplicate rows:", df.duplicated().sum())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Data cleaning

# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing categorical values with the most common value
for col in ['gender', 'country', 'referral_channel']:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Dataset after cleaning:", df.shape)

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
# Fill missing values in region
df['region'] = df['region'].fillna(df['region'].mode()[0])

print("Dataset shape:", df.shape)

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
print(df.shape)
```

```python
# STEP 2: Feature Encoding

df_encoded = df.copy()

# Convert categorical columns into numerical columns
categorical_columns = [
    'client_type',
    'gender',
    'country',
    'region',
    'acquisition_purpose',
    'loan_applied',
    'referral_channel'
]

df_encoded = pd.get_dummies(
    df_encoded,
    columns=categorical_columns,
    drop_first=False,
    dtype=int
)

print("Encoding completed!")
print("Encoded dataset shape:", df_encoded.shape)

display(df_encoded.head())
```

```python
# STEP 3: Feature Scaling

from sklearn.preprocessing import StandardScaler

# Select numerical features for clustering
features = df_encoded.drop(columns=['client_id', 'date_of_birth'])

# Convert date of birth into a numerical age
df_encoded['age'] = 2026 - pd.to_datetime(df_encoded['date_of_birth']).dt.year

# Recreate features after adding age
features = df_encoded.drop(columns=['client_id', 'date_of_birth'])

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(features)

print("Feature scaling completed!")
print("Scaled data shape:", X_scaled.shape)
```

```python
# STEP 4: Find the optimal number of clusters using the Elbow Method

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), inertia, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal Number of Clusters')
plt.show()
```

```python
# STEP 5: Calculate Silhouette Scores

from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)
    print(f"K = {k}, Silhouette Score = {score:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score for Different Cluster Numbers')
plt.show()
```

```python
# STEP 6: Apply K-Means with 5 clusters

kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)

# Create cluster labels
df['cluster'] = kmeans.fit_predict(X_scaled)

print("K-Means clustering completed!")
print("\nNumber of customers in each cluster:")
print(df['cluster'].value_counts().sort_index())
```

```python
# STEP 7: Analyse the buyer clusters

cluster_summary = df.groupby('cluster').agg({
    'client_type': lambda x: x.mode()[0],
    'gender': lambda x: x.mode()[0],
    'country': lambda x: x.mode()[0],
    'region': lambda x: x.mode()[0],
    'acquisition_purpose': lambda x: x.mode()[0],
    'loan_applied': lambda x: x.mode()[0],
    'referral_channel': lambda x: x.mode()[0],
    'satisfaction_score': 'mean'
}).round(2)

cluster_summary['buyer_count'] = df['cluster'].value_counts().sort_index()

display(cluster_summary)
```

```python
# Print the complete cluster summary clearly

print(cluster_summary.to_string())
```

```python
# STEP 8: Hierarchical Clustering

from sklearn.cluster import AgglomerativeClustering

hierarchical = AgglomerativeClustering(n_clusters=5)

hierarchical_labels = hierarchical.fit_predict(X_scaled)

df['hierarchical_cluster'] = hierarchical_labels

print("Hierarchical clustering completed!")
print("\nNumber of buyers in each hierarchical cluster:")
print(df['hierarchical_cluster'].value_counts().sort_index())
```

```python
# STEP 9: Visualize K-Means Buyer Clusters

from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=df['cluster'],
    alpha=0.7
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('K-Means Buyer Segmentation')
plt.colorbar(scatter, label='Cluster')
plt.show()
```

```python
# STEP 10: Final Buyer Segment Profiles

profile = df.groupby('cluster').agg(
    buyer_count=('client_id', 'count'),
    average_satisfaction=('satisfaction_score', 'mean'),
    common_client_type=('client_type', lambda x: x.mode()[0]),
    common_gender=('gender', lambda x: x.mode()[0]),
    common_country=('country', lambda x: x.mode()[0]),
    common_region=('region', lambda x: x.mode()[0]),
    common_purpose=('acquisition_purpose', lambda x: x.mode()[0]),
    common_loan_status=('loan_applied', lambda x: x.mode()[0]),
    common_referral=('referral_channel', lambda x: x.mode()[0])
).round(2)

display(profile)
```

```python
# STEP 11: Cluster Size Visualization

import matplotlib.pyplot as plt

cluster_counts = df['cluster'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(cluster_counts.index.astype(str), cluster_counts.values)

plt.xlabel('Buyer Cluster')
plt.ylabel('Number of Buyers')
plt.title('Number of Buyers in Each K-Means Cluster')

plt.show()
```

```python
# STEP 12: Compare K-Means and Hierarchical Clustering

comparison = pd.DataFrame({
    'K-Means': df['cluster'].value_counts().sort_index(),
    'Hierarchical': df['hierarchical_cluster'].value_counts().sort_index()
})

comparison.index.name = 'Cluster'

display(comparison)
```

```python
# STEP 13: Final Results Table

final_results = profile.copy()

final_results['segment_name'] = [
    'Small International Investors',
    'Loan-Financed UK Investors',
    'Mainstream Indian Investors',
    'Corporate Indian Investors',
    'High-Satisfaction Dubai Investors'
]

final_results = final_results[
    [
        'segment_name',
        'buyer_count',
        'average_satisfaction',
        'common_client_type',
        'common_gender',
        'common_country',
        'common_region',
        'common_purpose',
        'common_loan_status',
        'common_referral'
    ]
]

display(final_results)
```

```python
# STEP 14: Business Recommendations for Each Buyer Segment

recommendations = {
    'Small International Investors':
        'Use targeted online-search campaigns and provide personalized investment property options.',

    'Loan-Financed UK Investors':
        'Promote loan-friendly investment properties and advertise through property portals.',

    'Mainstream Indian Investors':
        'Focus on affordable investment opportunities and strengthen online-search marketing.',

    'Corporate Indian Investors':
        'Offer commercial and bulk investment opportunities and use social-media campaigns.',

    'High-Satisfaction Dubai Investors':
        'Promote premium investment properties and financing options through social media.'
}

final_results['recommendation'] = final_results['segment_name'].map(recommendations)

display(final_results[['segment_name', 'buyer_count',
                       'average_satisfaction', 'recommendation']])
```
